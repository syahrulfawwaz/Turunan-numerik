**TURUNAN NUMERIK**

```python
import math

print("="*60)
print(" -<IMPLEMENTASI>- ")
print("="*60)

def f(x):
    return 2 * x * math.cos(2 * x)

x = 2.0
h = 0.1
x_analitik = 4.7

turunan_maju = (f(x + h) - f(x)) / h

turunan_mundur = (f(x) - f(x - h)) / h

turunan_tengah = (f(x + h) - f(x - h)) / (2 * h)

print(f"Nilai x = {x}")
print(f"Nilai h = {h}")
print("="*50)
print(f"Hasil Analitik = {x_analitik}")
print(f"Beda Maju = {turunan_maju}")
print(f"Beda Mundur = {turunan_mundur}")
print(f"Beda Tengah = {turunan_tengah}")
```
